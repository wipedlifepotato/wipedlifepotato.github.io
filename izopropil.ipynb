Мы взяли для этого 15 мл конц соляной кислоты и 15 мл абсолютированного изопропилового. Грелось при 60 градусов на водяной бане, потом осталось на ночь. Тк у нас растворитель тут полярный протонный (вода), то шло по Sn1 пути, который очень медленный. Собираюсь попробовать добавить хлорид цинка, это добавит ион хлора и к тому же оно ведет себя как кислота льюиса и принимает электронную пару, тем самым проще забирая -OH.
R-OH + H+ => R-OH2(+) далее у нас идет Sn1, R(+) + HOH, а туда уже цепляется оставшийся Cl. Если добавить туда щелочи или воды будет много, то может пойти обратный процесс, но этому мешает то, что RCl плохо растворимо в воде. Тем самым оно образует слой или улетучивается.
График тут можно нарисовать процесса, но он низкий. Условимся что там 1 мл слоя.
Тогда для данных, для numpy у нас где 60 это градусы, а 1 это сколько времени кипятилось, а 0 это сколько хлорида цинка
15:15:60:1 = 1
