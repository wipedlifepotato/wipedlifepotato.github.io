```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import matplotlib.dates as mdates
# data, effect, activity, toxic
data_before_tea = [
    ("15:55", 0, 1, 1),
    ("16:06", 0, 5, 1),
    ("17:00", 3, 0, 5),
    ("17:39", 5, 0, 5)
]
#print( [d[0] for d in data] )
times = [d[0] for d in data_before_tea]
data = np.array([d[1:] for d in data_before_tea], dtype=np.int32)
df = pd.DataFrame(data, index=pd.to_datetime(times), columns=["эффект", "активность", "побочные_эффекты"])
df = df.sort_index()

fig, axes = plt.subplots(nrows=3, ncols=1, sharex=True, figsize=(10, 7))
fig.suptitle("Динамика: эффект, активность и токсичность во времени (паслён)", fontsize=14)

ax = axes[0]
ax.plot(df.index, df["эффект"], color="tab:blue", linewidth=1.5, label="Эффект")
ax.scatter(df.index, df["эффект"], color="tab:blue")
ax.set_ylabel("Эффект усталости (баллы)")

ax = axes[1]
ax.plot(df.index, df["активность"], color="tab:green", linewidth=1.5, label="Активность")
ax.scatter(df.index, df["активность"], color="tab:green")
ax.set_ylabel("Эффект возбуждения")

ax = axes[2]
ax.plot(df.index, df["побочные_эффекты"], color="tab:red", linewidth=1.5, label="Токсичность")
ax.scatter(df.index, df["побочные_эффекты"], color="tab:red")
ax.set_ylabel("побочные эффекты (баллы)")

for ax in axes:
    ax.xaxis.set_major_formatter(mdates.DateFormatter('%H:%M'))
fig.autofmt_xdate()

plt.show()

#fig, axes = plt.subplots(nrows=3,ncols=1,sharex=True, figsize=(10,10))
```
