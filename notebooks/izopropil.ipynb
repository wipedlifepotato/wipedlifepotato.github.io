```python
from molmass import Formula
import rdkit
from rdkit import Chem
from rdkit.Chem import AllChem
from rdkit.Chem.Draw import IPythonConsole
IPythonConsole.molSize = 250,250
m = Chem.MolFromSmiles('CC(O[H])C')
AllChem.ComputeGasteigerCharges(m)
m
```

![](izopropil/2026_04_05.jpg)

```python
m2 = Chem.Mol(m)
for at in m2.GetAtoms():
    lbl = '%.2f'%(at.GetDoubleProp("_GasteigerCharge"))
    at.SetProp('atomNote',lbl)
m2
```

```python
Izopropil = Formula("CH3CHOHCH3")
m = Chem.MolFromSmiles('CC(Cl)C')
AllChem.ComputeGasteigerCharges(m)

m
```

```python
m2 = Chem.Mol(m)
for at in m2.GetAtoms():
    lbl = '%.2f'%(at.GetDoubleProp("_GasteigerCharge"))
    at.SetProp('atomNote',lbl)
m2
```

```python
print(f"Масса изопропилового спирта: {Izopropil.mass}")
print(f"Масса изопропилхлорида: {Formula("CH3CHClCH3").mass}")
print("Изопропиловый спирт намного легче по плотности, он должен быть выше, тогда мы слой не там искали вообще")
```

Тк у нас растворитель тут полярный протонный (вода), то шло по Sn1 пути, который очень медленный. Собираюсь попробовать добавить хлорид цинка, это добавит ион хлора и к тому же оно ведет себя как кислота льюиса и принимает электронную пару, тем самым проще забирая -OH.
R-OH + H+ => R-OH2(+) далее у нас идет Sn1, R(+) + HOH, а туда уже цепляется оставшийся Cl. Если добавить туда щелочи или воды будет много, то может пойти обратный процесс, но этому мешает то, что RCl плохо растворимо в воде. Тем самым оно образует слой или улетучивается.
График тут можно нарисовать процесса, но он низкий. Условимся что там 1 мл слоя.
Тогда для данных, для numpy у нас где 60 это градусы, а 1 это сколько времени кипятилось, а 0 это сколько хлорида цинка
15:15:60:1 = 1
